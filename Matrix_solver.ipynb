```python
def to_subscript(number):
    """Converts an integer or string to its Unicode subscript representation."""
    subscript_map = {
        '0': '₀', '1': '₁', '2': '₂', '3': '₃', '4': '₄',
        '5': '₅', '6': '₆', '7': '₇', '8': '₈', '9': '₉'
    }
    return ''.join(subscript_map.get(digit, digit) for digit in str(number))
```

```python
def type_of_matrix(matrix):
  rows, cols = matrix.shape

  print(f"\nAnalyzing a {rows}x{cols} matrix:")

  # 1. Primary shape classification
  is_square = (rows == cols)
  if is_square:
    print("Matrix is a square matrix.")
  elif rows == 1:
    print("Matrix is a row vector.")
  elif cols == 1:
    print("Matrix is a column vector.")
  else:
    print("Matrix is a rectangular matrix.")

  # 2. Specific property checks using basic NumPy

  # Zero Matrix
  if np.all(matrix == 0):
    print("  - It is a Zero matrix.")

  if is_square: # Properties specifically for square matrices
    # Diagonal Matrix
    is_diagonal = np.all(matrix == np.diag(np.diag(matrix)))
    if is_diagonal:
      print("  - It is a Diagonal matrix.")

      # Scalar Matrix (subset of Diagonal)
      # A scalar matrix has all diagonal elements equal. If it's a non-zero diagonal matrix with equal diagonal elements, it's scalar.
      # A zero matrix is also scalar, but 'Zero matrix' is a more specific classification.
      if np.unique(np.diag(matrix)).size == 1:
        print("  - It is a Scalar matrix.")

        # Identity Matrix (subset of Scalar, where diagonal elements are 1)
        if np.array_equal(matrix, np.identity(rows)):
          print("  - It is an Identity matrix.")

    # Symmetric Matrix
    if np.array_equal(matrix, matrix.T):
      print("  - It is a Symmetric matrix.")

    # Skew-Symmetric Matrix
    if np.array_equal(matrix, -matrix.T):
      print("  - It is a Skew-Symmetric matrix.")

    # Orthogonal Matrix
    # Using np.allclose for floating point comparisons due to potential precision issues.
    if np.allclose(np.dot(matrix, matrix.T), np.identity(rows)):
      print("  - It is an Orthogonal matrix.")
```

### Symbolic Matrix Equality and Solving for Variables

To solve for unknown variables (like `x` or `y`) in a matrix by equating it with another, we'll use `sympy` to handle symbolic expressions. Let's redefine the second matrix to allow for symbolic entries.

```python
def matrix_two_for_compare():
  print(matrix)
  # Define symbolic variables
  x, y, z = sympy.symbols('x y z')

  # Re-enter elements for the second matrix, allowing symbolic expressions
  rows2_sym = matrix.shape[0] # Use dimensions from the first matrix
  columns2_sym = matrix.shape[1]

  matrix_elements_sym2 = []
  print(f"\nEnter elements for the second symbolic matrix (dimensions: {rows2_sym}x{columns2_sym}). You can use 'x', 'y', 'z', or numerical values:")
  for r in range(rows2_sym):
    for c in range(columns2_sym):
      # Use eval to convert string input to a sympy expression
      val_str = input(f"Enter expression for C{to_subscript(r)}{to_subscript(c)}: ")
      matrix_elements_sym2.append(sympy.sympify(val_str))

  # Create the sympy Matrix
  matrix_sym2 = sympy.Matrix(matrix_elements_sym2).reshape(rows2_sym, columns2_sym)
  print("\nYour second symbolic matrix:")
  display(matrix_sym2)
  return matrix_sym2
```

### Set up and Solve the Symbolic Equation

Now, we'll equate the first numerical matrix (converted to a `sympy` matrix) with the second symbolic matrix and solve for any unknown variables.

```python
# Convert the numerical numpy matrix to a sympy Matrix for comparison
def Compare_Matrices():
  matrix_sym1 = sympy.Matrix(matrix) # This uses the global 'matrix'
  matrix_sym2_received = matrix_two_for_compare() # Call the function, get the returned symbolic matrix

  # Get dimensions from the received symbolic matrix
  rows2_sym, columns2_sym = matrix_sym2_received.shape

  # Create the equation: matrix_sym1 = matrix_sym2_received
  equations = []
  for r in range(rows2_sym):
      for c in range(columns2_sym):
          equations.append(sympy.Eq(matrix_sym1[r, c], matrix_sym2_received[r, c]))

  print("Equations derived from matrix equality:")
  for eq in equations:
      print(eq)

  # Extract all symbols from the equations to solve for
  all_symbols = set()
  for eq in equations:
      all_symbols.update(eq.free_symbols)

  if all_symbols:
      solution = sympy.solve(equations, list(all_symbols))
      if solution:
          print("\nSolution for the variables:")
          display(solution)
      else:
          print("\nNo consistent solution found for the variables.")
  else:
      print("\nNo symbolic variables found in the second matrix to solve for, or the matrices are already fully defined numerically.")
      # If no symbols, check if the numerical matrices are equal
      # Need to convert sympy Matrix to numpy array for np.array_equal
      if np.array_equal(matrix, np.array(matrix_sym2_received).astype(int)):
          print("The two matrices are numerically equal.")
      else:
          print("The two matrices are numerically NOT equal.")
```

```python
def symmetric_skew_symmetric(matrix):
    if matrix.shape[0] != matrix.shape[1]:
        print("Matrix must be square.")
        return

    transpose = matrix.T

    symmetric = (matrix + transpose) / 2
    skew_symmetric = (matrix - transpose) / 2

    print("Original Matrix:")
    print(matrix)

    print("\nSymmetric Matrix:")
    print(symmetric)

    print("\nSkew-Symmetric Matrix:")
    print(skew_symmetric)

    print("\nSum:")
    print("symmetric + skew_symmetric")
```

```python
def Area_of_traingle():
  print("(x1,y1) , (x2,y2) and (x3,y3) are points of triangles.")
  x1=int(input("Enter the value of x1: "))
  y1=int(input("Enter the value of y1: "))
  x2=int(input("Enter the value of x2: "))
  y2=int(input("Enter the value of y2: "))
  x3=int(input("Enter the value of x3: "))
  y3=int(input("Enter the value of y3: "))
  # Corrected matrix construction: y3 instead of y2
  a=np.array([[x1,y1,1],[x2,y2,1],[x3,y3,1]])
  # Corrected area calculation: take absolute value of determinant
  area=abs(np.linalg.det(a))/2
  print("Area of triangle is: ",area)
```

```python
def minor(matrix, row, column):

    minor_matrix = np.delete(matrix, row, axis=0)
    minor_matrix = np.delete(minor_matrix, column, axis=1)

    return np.linalg.det(minor_matrix)
```

```python
def minor_matrix(matrix):

    rows, columns = matrix.shape
    result = np.zeros((rows, columns))

    for i in range(rows):
        for j in range(columns):
            result[i, j] = minor(matrix, i, j)

    return result
```

```python
def cofactor(matrix, row, column):

    M = minor(matrix, row, column)

    return (-1) ** (row + column) * M
```

```python
def cofactor_matrix(matrix):

    rows, columns = matrix.shape
    result = np.zeros((rows, columns))

    for i in range(rows):
        for j in range(columns):
            result[i, j] = cofactor(matrix, i, j)

    return result
```

```python
def Adjoint(matrix):
  matrix1=cofactor_matrix(matrix)
  return matrix1.T
```

```python
def Inverse(matrix):


    if matrix.shape[0] != matrix.shape[1]:
        print("Matrix must be square.")
        return

    determinant = np.linalg.det(matrix)

    if np.isclose(determinant, 0):
        print("Matrix is not invertible.")
        return

    inverse_matrix = Adjoint(matrix) / determinant

    print("\nResult of Inverse:")
    print(inverse_matrix)
```

```python
def system_of_equations():
  variables=int(input("Enter the number of equations and number of variables in one equation (both are same): "))
  A = []
  for r in range(variables):
    for c in range(variables):
      n = int(input(f"Enter number for C{to_subscript(r)}{to_subscript(c)}: "))
      A.append(n)
  B=[]
  for i in range(variables):
    n = int(input(f"Enter number for B{to_subscript(i)}: "))
    B.append(n)
  A = np.array(A).reshape(variables, variables)

  solution = np.linalg.solve(A, B)

  print("Solution:")
  print(solution)
```

```python
def main_menu():
    global matrix # Declare matrix as global to allow modification
    while True:
        print("\n" + "="*20 + " MATRIX SOLVER MENU " + "="*20)
        print("1. Create Matrix")
        print("2. Matrix Classification")
        print("3. Addition")
        print("4. Subtraction")
        print("5. Multiplication")
        print("6. Scalar Multiplication")
        print("7. Transpose")
        print("8. Determinant")
        print("9. Minors & Cofactors (individual element)")
        print("10. Minor Matrix")
        print("11. Cofactor Matrix")
        print("12. Adjoint")
        print("13. Inverse")
        print("14. System of Equations")
        print("15. Rank")
        print("16. Area of Triangle (from coordinates)")
        print("17. Symmetric & Skew-Symmetric Decomposition")
        print("18. Compare Matrices (Symbolic)")
        print("19. Exit")
        print("="*60)

        choice = input("Enter your choice: ")

        if choice == '1':
            create_matrix_interactive()
        elif matrix is None and choice not in ['1', '14', '16', '19']:
            print("Please create a matrix first using option 1, or choose System of Equations (14) / Area of Triangle (16).")
        elif choice == '2':
            type_of_matrix(matrix)
        elif choice == '3':
            add_matrices_numerical()
        elif choice == '4':
            subtract_matrices_numerical()
        elif choice == '5':
            multiply_matrices_numerical()
        elif choice == '6':
            scalar_multiplication_numerical()
        elif choice == '7':
            Transpose()
        elif choice == '8':
            Determinant()
        elif choice == '9':
            if matrix.shape[0] != matrix.shape[1]:
                print("Matrix must be square to calculate minor/cofactor of an individual element.")
                continue
            r_idx = int(input("Enter row index for minor/cofactor (0-indexed): "))
            c_idx = int(input("Enter column index for minor/cofactor (0-indexed): "))
            print(f"Minor of element ({r_idx}, {c_idx}): {minor(matrix, r_idx, c_idx)}")
            print(f"Cofactor of element ({r_idx}, {c_idx}): {cofactor(matrix, r_idx, c_idx)}")
        elif choice == '10':
            if matrix.shape[0] != matrix.shape[1]:
                print("Matrix must be square to calculate the Minor Matrix.")
                continue
            print("Minor Matrix:")
            print(minor_matrix(matrix))
        elif choice == '11':
            if matrix.shape[0] != matrix.shape[1]:
                print("Matrix must be square to calculate the Cofactor Matrix.")
                continue
            print("Cofactor Matrix:")
            print(cofactor_matrix(matrix))
        elif choice == '12':
            Adjoint(matrix)
        elif choice == '13':
            Inverse(matrix)
        elif choice == '14':
            system_of_equations()
        elif choice == '15':
            Rank()
        elif choice == '16':
            Area_of_traingle()
        elif choice == '17':
            symmetric_skew_symmetric(matrix)
        elif choice == '18':
            Compare_Matrices()
        elif choice == '19':
            print("Exiting Matrix Solver. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 19.")

# Call the main menu function to start the program
main_menu()
```
